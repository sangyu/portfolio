<div class="home-about">
<p>I'm a computational behavior scientist modeling complex behavior and physiology systems. I specialize in <a href="neuroscience.html">neuromodulation</a>, longitudinal behavior and physiology tracking, <a href="blog/03_itc.html">learning mechanisms</a>, <a href="statistics.html">estimation statistics</a>, large-dimensional data analysis and <a href="blog/09_whorlmap.html">visualization</a>, reproducible analysis workflows, and I think about <a href="blog/04_llm_evals.html">AI agent workflows and evaluation</a>. My work asks how we infer hidden states from noisy, high-dimensional evidence. Hunger, satiety, impulsivity, attention, social influence, and model capability are not directly visible.</p>
<p>I am currently a Senior Research Fellow at the <a href="https://www.duke-nus.edu.sg/nbd/the-team/faculty-members/details/Detail/adam-claridge-chang" target="_blank" rel="noopener">Adam Claridge-Chang Lab</a>. Previously I was a PhD student at the <a href="https://tonegawalab.mit.edu/" target="_blank" rel="noopener">Tonegawa Lab</a> at MIT.</p>
