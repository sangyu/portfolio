## Four Agents

I built a manuscript evaluation pipeline for scientific writing — specifically the kind of high-dimensional behavioral paper I work on. Four role-based LLM subagents critique a manuscript and each other.

A biologist reads the manuscript and proposes interpretations: what is the most plausible story given these results? A bioinformatician challenges that: are there confounds? Is the claim supported by the evidence actually presented? A reviewer scores against an explicit rubric — novelty, rigor, clarity, impact. An editor synthesizes everything into a revision letter with ranked action items.

The agents run in sequence; each one sees what the previous ones wrote. I've been running this on drafts from our group. The bioinformatician is annoyingly right most of the time.

## No Ground Truth

The build was the easy part.

For a chess engine, ground truth is easy: did it win? For manuscript quality, there is no ground truth. Expert opinions don't agree. Rubrics can be applied well or badly. "Scientifically rigorous and clearly communicated" is not a function you can compute. It is an inferred state.

This is the same measurement problem I work on with behavioral states. Hunger, satiety, impulsivity — not directly observable. Inferred from noisy, high-dimensional evidence. The key question is always the same: have you actually shifted the state, or just one metric?

The evaluation framework I'm building borrows from DESTRA. Compare the LLM judges against human expert reviewers, estimate inter-rater agreement, and use that to distinguish "the agent is identifying real problems" from "the agent is generating plausible-sounding criticism." The bootstrap inter-rater metric from DABEST is doing the same job here as it does in the animal work — quantifying agreement in the presence of imperfect reference standards on both sides.

The measurement problem is the same problem.
